```python
import pandas as pd
data = pd.read_csv("dataset_1.csv", encoding="utf-8")
```

```python
#task1
# Lớn nhất
data["Population"] = data["Population"].str.replace(",", "")
data["Population"] = pd.to_numeric(data["Population"])
top10_population = data.sort_values(by="Population", ascending=False).head(10)

# Nhỏ nhất
bottom10_population = data.sort_values(by="Population", ascending=True).head(10)

print("Top 10 thành phố có dân số lớn nhất:")
print(top10_population[['City', 'Population']])

print("\nTop 10 thành phố có dân số nhỏ nhất:")
print(bottom10_population[['City', 'Population']])
```

```python
#task2
countries_with_3_or_more = data['Country'].value_counts()
countries_with_3_or_more = countries_with_3_or_more[countries_with_3_or_more >= 3].index.tolist()
print("Các quốc gia có ít nhất 3 thành phố trong danh sách:")
print(countries_with_3_or_more)
```

```python
#task3
top5_countries = data['Country'].value_counts().head(5)
print("Top 5 quốc gia có nhiều thành phố nhất:")
print(top5_countries)
```

```python
#task4
top20_population = data.sort_values(by="Population", ascending=False).head(20)
top20_area = data.sort_values(by="Area KM2", ascending=False).head(20)

# Giao nhau theo tên thành phố
top_both = pd.merge(top20_population, top20_area, on="City")
print("Các thành phố có dân số & diện tích đều thuộc top 20:")
print(top_both[['City', 'Population_x', 'Area KM2_x']])

```

```python
#task5
data["Density KM2"] = data["Density KM2"].str.replace(",", "")
data["Density KM2"] = pd.to_numeric(data["Density KM2"])
density_by_country = data.groupby("Country")["Density KM2"].mean().sort_values(ascending=False)
print("Mật độ dân số trung bình theo quốc gia (tính theo km2):")
print(density_by_country)
```

```python
#task6
# Đếm số thành phố mỗi quốc gia
city_counts = data['Country'].value_counts()
eligible_countries = city_counts[city_counts >= 2].index

# Lọc các quốc gia hợp lệ
filtered_df = data[data['Country'].isin(eligible_countries)]

# Tìm thành phố đông dân nhất trong từng quốc gia
top_city_each_country = filtered_df.sort_values(by='Population', ascending=False).groupby('Country').first()

print("Thành phố đông dân nhất ở mỗi quốc gia có từ 2 thành phố trở lên:")
print(top_city_each_country[['City', 'Population']])
```
