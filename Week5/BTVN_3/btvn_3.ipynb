```python
import pandas as pd
data = pd.read_csv("dataset_3.csv")

# Task 1: Liệt kê các sản phẩm có doanh thu > 100.000 EUR
data["revenue"] = data["price"] * data["units_sold"]
revenue_over_100k = data[data["revenue"] > 100000]
print("🔹 Task 1 - Sản phẩm doanh thu > 100.000 EUR:")
print(revenue_over_100k[["title", "price", "units_sold", "revenue"]].head())

# Task 2: Sản phẩm top 100 units_sold và thuộc top 10% rating cao nhất
top100_units_sold = data.nlargest(100, "units_sold")
rating_threshold = data["rating"].quantile(0.9)
top_combined = top100_units_sold[top100_units_sold["rating"] >= rating_threshold]
print("\n🔹 Task 2 - Sản phẩm top 100 units_sold và rating trong top 10%:")
print(top_combined[["title", "units_sold", "rating"]].head())

# Task 3: Kiểm tra tương quan giữa rating_count và units_sold
correlation = data[["rating_count", "units_sold"]].corr().iloc[0, 1]
print(f"\n🔹 Task 3 - Hệ số tương quan giữa rating_count và units_sold: {correlation:.4f}")

# Task 4: Top 50 sản phẩm rating từ 3.0 đến 4.0 bán chạy nhất
filtered = data[(data["rating"] >= 3.0) & (data["rating"] <= 4.0)]
top50_in_range = filtered.nlargest(50, "units_sold")
print("\n🔹 Task 4 - Top 50 sản phẩm rating từ 3.0 đến 4.0 bán chạy nhất:")
print(top50_in_range[["title", "rating", "units_sold"]].head())

# Task 5: Đếm số sản phẩm có giá > 50 EUR
count_price_over_50 = data[data["price"] > 50].shape[0]
print(f"\n🔹 Task 5 - Số sản phẩm có giá > 50 EUR: {count_price_over_50}")
```
