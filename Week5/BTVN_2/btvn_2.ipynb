```python
import pandas as pd
data = pd.read_csv("dataset_2.csv", encoding="utf-8")
```

```python

# Task 1
data["average_score"] = data[["math score", "reading score", "writing score"]].mean(axis=1)
top_math = data["math score"].nlargest(100).index
top_reading = data["reading score"].nlargest(100).index
top_both = top_math.intersection(top_reading)[:10]
print("Index 10 học sinh (math & reading top 100):", list(top_both))
```

```python
# Task 2
top20_writing_by_group = data.groupby("race/ethnicity").apply(
    lambda g: g.nlargest(20, "writing score")
).reset_index(drop=True)
print("Index top 20 writing theo group:", list(top20_writing_by_group.index[:10]))
```

```python
# Task 3: Thống kê số lượng học sinh group A theo parental level of education
group_a_counts = data[data["race/ethnicity"] == "group A"]["parental level of education"].value_counts()
print("Task 3 - Số lượng học sinh group A theo học vấn cha mẹ:\n", group_a_counts)
```

```python
# Task 4: Trung bình điểm theo học vấn cha mẹ
avg_score_by_edu = data.groupby("parental level of education")["average_score"].mean().sort_values(ascending=False)
print("Trung bình điểm theo học vấn cha mẹ:\n", avg_score_by_edu)
```

```python
# Task 5: Trung bình điểm theo loại bữa ăn
avg_score_by_lunch = data.groupby("lunch")["average_score"].mean()
print("Task 5 - Trung bình điểm theo lunch:\n", avg_score_by_lunch)
```

```python
# Task 6: Top 10 học sinh có điểm toán cao nhất của mỗi group
top10_math_by_group = data.groupby("race/ethnicity").apply(
    lambda g: g.nlargest(10, "math score")
).reset_index(drop=True)
print("Top 10 học sinh toán cao nhất mỗi group:\n", top10_math_by_group[["race/ethnicity", "math score"]].head())
```

```python
# Task 7: So sánh điểm trung bình theo group + học vấn + lunch + test prep
comparison = data.groupby(
    ["race/ethnicity", "parental level of education", "lunch", "test preparation course"]
)["average_score"].mean().unstack()
print("So sánh điểm theo việc tham gia ôn tập :\n", comparison)
```
