```python
import pandas as pd
data = pd.read_csv("Week4_1_data.csv")
data["educational_level"].value_counts()
```

```python
#task2
top_20_income = data.sort_values(by='annual_income', ascending=False).head(20)

print(top_20_income)
```

```python
#task3
filtered = data[(data['year_of_birth'] > 1960) & (data['annual_income'] > 50000)]
print(filtered)
```

```python
#task4
filtered_top20 = data[(data['year_of_birth'] > 1960) & (data['annual_income'] > 50000)].sort_values(by='annual_income', ascending=False).head(20)
print(filtered_top20)
```

```python
#task5
filtered_marital_status = data[data['marital_status'].isin(['Married', 'Divorced'])]
print(filtered_marital_status)
```

```python
#Task6
avg_income_by_edu = data.groupby('educational_level')['annual_income'].mean().reset_index()
print(avg_income_by_edu)
```

```python
#task7
avg_income_by_edu_and_marital_status = data.groupby(['educational_level', 'marital_status'])['annual_income'].mean().reset_index()
print(avg_income_by_edu_and_marital_status)
```
