```python
#1b
import pandas as pd
data = pd.read_csv("Iris (1).csv")
for species in data["Species"].unique():
    #lay 10 mau hoa dau tien moi loai
    subset = data[data["Species"] == species].head(10)
    #print(subset)
    max_sepal_length = subset["SepalLengthCm"].max()
    print(f"{species}: {max_sepal_length}")
```

```python
#2b
data['max_sepal_length'] = data["SepalWidthCm"] + data["SepalLengthCm"]
top_50 = data.sort_values(by='max_sepal_length', ascending=False).head(50)
print(top_50["Species"].value_counts())

```

```python
#3b - a
import numpy as np
length_mm = float(input("Nhập chiều dài đài hoa (mm): "))
width_mm = float(input("Nhập chiều rộng đài hoa (mm): "))

length_cm = length_mm/10
width_cm = width_mm/10

#Tinh khoang cach den tưng bong hoa
#|x - y| = sqrt((x1 - x2)^2 + (y1-y2)^2)
data["distance"] = np.sqrt((data["SepalLengthCm"] - length_cm)**2 + (data["SepalWidthCm"] - width_cm)**2)
data_dis_sorted = data.sort_values(by='distance')
result_species = None
for i in range(len(data_dis_sorted)):
    curr_dis = data_dis_sorted.iloc[i]['distance']
    curr_species = data_dis_sorted.iloc[i]['Species']
    if i + 1 < len(data_dis_sorted):
        next_dis = data_dis_sorted.iloc[i+1]['distance']
        next_species = data_dis_sorted.iloc[i+1]['Species']
        if np.isclose(curr_dis, next_dis) and curr_species != next_species:
            continue
        else:
            result_species = curr_species
            break
    else:
        result_species = curr_species

print(f"Dự đoán là hoa: {result_species}")

```
